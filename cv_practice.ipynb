```python
from torch.utils.data import TensorDataset


x = torch.tensor([
    [1,2],
    [3,4],
    [5,6],
    [7,8]
])

y = torch.tensor([0,1,0,1])

data = TensorDataset(x,y)

print(len(data))
print(data[0])
```

```python
from torch.utils.data import DataLoader

Loader = DataLoader(
    data,
    batch_size=2,
    shuffle=True
)

for x_batch,y_batch in Loader:
  print(f"x_batch is {x_batch}")
  print(f"y_batch is {y_batch}")
```

```python
# ### Important Terms

# Dataset → collection of individual samples
# DataLoader → creates batches from the Dataset
# Batch → group of samples processed together
# Iteration/Step → one batch processed
# Epoch → entire dataset processed once


# Example:
# Dataset = 1000 images
# Batch size = 100

# Iterations per epoch = 1000 / 100 = 10
```

```python
# model building
import torch.nn as nn
class simpleModal(nn.Module):

  def __init__(self):
    super().__init__()

    self.layer = nn.Linear(2,1)

  def forward(self,x):
    return self.layer(x)

model = simpleModal()
print(model)
```

```python
print(model.layer.weight)
print(model.layer.bias)
```

```python
# giving x values into model

x = torch.tensor([5.0,2.0])
output=model(x)

print(output)
```

```python
import torch
import torch.nn as nn

# Model
model = nn.Sequential(
    nn.Linear(2, 4),
    nn.ReLU(),
    nn.Linear(4, 1)
)

# Input and target
x = torch.tensor([[5.0, 2.0]])
target = torch.tensor([[4.0]])

# Forward pass
output = model(x)

# Loss
loss_fn = nn.MSELoss()
loss = loss_fn(output, target)

# Backpropagation
loss.backward()

# Gradients
print("Weight gradients:")
print(model[0].weight.grad)

print("\nBias gradients:")
print(model[0].bias.grad)
```

```python
tar = torch.tensor([4.0])

loss_fn = nn.MSELoss()
loss=loss_fn(output,tar)
print(loss)
```
